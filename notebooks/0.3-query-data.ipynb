# Buscar datos

Este cuaderno te permite buscar en las importaciones y exportaciones de Colombia (2001 a hoy) sin necesidad de programar. Cambia los valores de la celda de **Configuración** y ejecuta el cuaderno completo.

## Cómo usarlo

1. En la celda de **Configuración**, completa lo que quieras buscar (abajo explicamos cada opción).
2. Ejecuta todas las celdas del cuaderno, en orden.
3. El resultado aparece debajo de la consulta y se guarda automáticamente en `outputs/tables/`.

## Qué puedes buscar

| Opción | Qué es | Ejemplo |
|---|---|---|
| `query_regimen` | `"i"` (importaciones) o `"e"` (exportaciones) | `"i"` |
| `query_years` | El año o los años a consultar | `2024` o `[2023, 2024, 2025]` |
| `query_nombre` *(opcional)* | Nombre (o parte del nombre) de la empresa colombiana que importa o exporta. En exportaciones solo existe desde 2007, y a veces figura como "PERSONA NATURAL": para no perder filas, conviene buscar por NIT. | `"acme"` |
| `query_codigo` *(opcional)* | Subpartida arancelaria NANDINA (10 dígitos), completa o solo el comienzo; los puntos son opcionales. Puede ser una lista | `"2701"`, `"27.01"` o `["2701", "2704"]` |
| `query_nit` *(opcional)* | NIT de la empresa, sin dígito de verificación; los puntos son opcionales. Puede ser una lista | `"890900608"` |
| `query_formato` | Formato del archivo: `"csv"` (por defecto), `"xlsx"` o `"parquet"` | `"csv"` |

Hace falta completar al menos `query_nombre`, `query_codigo` o `query_nit`.

**Columnas del resultado:** no aparecen las columnas que no tienen ningún dato en los años consultados (por ejemplo, las de los formatos viejos de la DIAN en una consulta de 2020 a 2026). Las columnas con datos, aunque sean todos 0, se mantienen: un 0 es un valor (por ejemplo, "no se cobró antidumping").

El resultado agrega las columnas `AÑO` y `MES`, tomadas de `PERIODO`.

**Nombre del archivo:** se arma con la búsqueda y la fecha y hora de exportación (por ejemplo, `co_importacion_2701_2024-2026_20261111T153012.csv`), así que nunca reemplaza un archivo anterior.

**Excel:** admite como máximo 1.048.576 filas por hoja; para búsquedas amplias usa `"csv"` o `"parquet"`.

````python
from colombia_dian_comex import Consulta, export_query_result, query_data, utils
````

````python
# --- Configuración ---
query_regimen = "i"  # "i" (importaciones) o "e" (exportaciones)
query_years = [2024, 2025, 2026]
query_nombre = None
query_codigo = "2701"  # Subpartida NANDINA o su comienzo; también una lista
query_nit = None
query_formato = "csv"  # "csv" (por defecto), "xlsx" o "parquet"

consulta = Consulta(
    years=query_years,
    tipo_regimen=query_regimen,
    nombre=query_nombre,
    codigo=query_codigo,
    nit=query_nit,
)
query_result = query_data(consulta)
display(query_result)

# Guarda el resultado en outputs/tables/
export_query_result(
    query_result,
    consulta,
    output_dir=utils.outputs_tables_dir(),
    formato=query_formato,
)
````
